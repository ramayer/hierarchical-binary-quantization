# Smoke test of HSQ quantizer.

```python
import hierarchical_binary_quantization.hbq as hbq
import torch
batches, vector_lenght = 3,8
z = torch.rand(batches,vector_lenght) * 2 - 1
z
```

```python
hbq.hbq(z,n_rounds=1)
```

```python
z = torch.rand(batches,16 * 16 * 3) * 2 - 1
hbq.hbq(z,n_rounds=2)
```

```python

x = torch.tensor([[-1.5, -0.5, 0.5, 1.5]], requires_grad=True)

quantizer = hbq.HBQQuantizer(n_rounds=2)
q_out, tokenids = quantizer(x)

loss = (q_out ** 2).sum()

# 4. Compute gradients
loss.backward()

# 5. Print results to verify
print("=== Validation Results ===")
print(f"Original Input x:         {x.data}")
print(f"Quantized Output q_out:  {q_out.data}")
print(f"Token IDs q_out:  {tokenids.data}")
print(f"Do gradients exist?      {x.grad is not None}")
print(f"Computed Gradients x.grad: {x.grad}")
```

```python
hbq_module = hbq.HBQQuantizer(1)
z = torch.rand(batches,3, 16, 16,requires_grad=True) * 2 - 1

q_out, indicies = hbq_module.forward(z)
q_out.shape, indicies.shape
```

```python
loss = (q_out ** 2).sum()
loss.backward()
```
